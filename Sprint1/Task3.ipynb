### Вариант задания


Создать систему учета животных в зоопарке. Базовый класс Animal описывает общие характеристики и поведение, а производные классы Dog, Cat и Bird представляют конкретные виды животных.

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

В проекте реализована система учета животных. Каждый объект имеет имя, возраст и вид. Для каждого животного предусмотрены базовые методы получения информации и приветствия. Производные классы добавляют уникальные свойства, такие как порода, цвет и размер крыла. В классе Zoo также реализован список животных, позволяющий добавлять и отображать информацию о всех обитателях.

Также используются конструкторы, свойства с геттерами и сеттерами, а также взаимодействие между объектами: зоопарк может добавлять животных, а сами животные могут общаться между собой.


#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

class Animal
{
    private string _name;
    private int _age;
    private string _species;

    public Animal()
    {
        _name = "Безымянный";
        _age = 0;
        _species = "Неизвестно";
    }

    public Animal(string name, int age, string species)
    {
        Name = name;
        Age = age;
        Species = species;
    }

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    public string Species
    {
        get { return _species; }
        set { _species = value; }
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Вид: {Species}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}, и я {Species}.");
    }
}

class Dog : Animal
{
    private string _breed;

    public Dog() : base()
    {
        Breed = "Неизвестно";
        Species = "Собака";
    }

    public Dog(string name, int age, string breed) : base(name, age, "Собака")
    {
        Breed = breed;
    }

    public string Breed
    {
        get { return _breed; }
        set { _breed = value; }
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Порода: {Breed}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Гав! Меня зовут {Name}. Я собака породы {Breed}.");
    }

    public void PlayWith(Animal animal)
    {
        Console.WriteLine($"{Name} играет с {animal.Name}.");
    }
}

class Cat : Animal
{
    private string _color;

    public Cat() : base()
    {
        Color = "Неизвестно";
        Species = "Кошка";
    }

    public Cat(string name, int age, string color) : base(name, age, "Кошка")
    {
        Color = color;
    }

    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Цвет: {Color}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Мяу! Я {Name}. Мой цвет: {Color}.");
    }
}

class Bird : Animal
{
    private string _wingSpan;

    public Bird() : base()
    {
        WingSpan = "Неизвестно";
        Species = "Птица";
    }

    public Bird(string name, int age, string wingSpan) : base(name, age, "Птица")
    {
        WingSpan = wingSpan;
    }

    public string WingSpan
    {
        get { return _wingSpan; }
        set { _wingSpan = value; }
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Размах крыльев: {WingSpan}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Чик-чирик! Я {Name}. Размах крыльев: {WingSpan}.");
    }
}

class Zoo
{
    private List<Animal> _animals = new List<Animal>();

    public void AddAnimal(Animal animal)
    {
        _animals.Add(animal);
    }

    public void ShowAnimals()
    {
        Console.WriteLine("Список животных в зоопарке:");
        foreach (var animal in _animals)
        {
            Console.WriteLine(animal.GetInfo());
            animal.SayHello();
            Console.WriteLine();
        }
    }
}

Dog dog = new Dog("Барбос", 5, "Овчарка");
Cat cat = new Cat("Мурка", 3, "Черная");
Bird bird = new Bird("Кеша", 1, "20 см");

Zoo zoo = new Zoo();
zoo.AddAnimal(dog);
zoo.AddAnimal(cat);
zoo.AddAnimal(bird);

dog.PlayWith(cat);
Console.WriteLine();
zoo.ShowAnimals();
```
